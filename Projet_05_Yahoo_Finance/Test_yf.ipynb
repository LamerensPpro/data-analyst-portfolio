```python
import yfinance as yf
from sqlalchemy import create_engine

tickers = ["TTE.PA", "MC.PA", "AIR.PA", "BNP.PA", "SAN.PA"]

df = yf.download(tickers, start="2024-01-01", end="2024-12-31")

df = df.stack(level=1)
df.columns.name = None   # ← enlève "Price"
df = df.reset_index()

# Vers PostgreSQL
engine = create_engine("postgresql://postgres:postgres@localhost:5432/stock_data")
df.to_sql('raw_stock_prices', engine, if_exists='replace', index=False)

print(f"✅ {len(df)} lignes")

```
